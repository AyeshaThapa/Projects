```python
#Display top 10 most frequent items in a column
if 'Item' in df.columns:
    top_10_items = df['Item'].value_counts().head(10)
    print("\nTop 10 Most Frequent Items:")
    print(top_10_items)
```

```python
#Plot a bar chart for top 1o items if "Item" column exists
if 'Item' in df.columns and 'price' in df.columns:
  top_items=df.groupby('Item')['price'].sum().sort_values(ascending=False).head(10)
  plt.figure(figsize=(10,6))
  sns.barplot(x=top_items.index,y=top_items.values,palette='viridis')
  plt.title('Top 10 Items by Revenue')
  plt.xlabel('Item')
  plt.ylabel('Revenue')
```
