```python
df.isnull().sum().sort_values(ascending=False)
```

Visualizing missing values

```python
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
```

Handling Missing Values (Mean Imputation)

```python
imputer = SimpleImputer(strategy="mean")

for col in df.columns:
    if df[col].isnull().sum() > 0 and df[col].dtype != 'object':
        df[col] = imputer.fit_transform(df[[col]])
```

Categorical Feature Analysis

Country Count

```python
df["Country"].nunique()
df["Status"].value_counts()
sns.countplot(x="Status", data=df)
plt.title("Developed vs Developing Countries")
plt.show()
```

Distribution of Life Expectancy

```python
plt.figure(figsize=(10,6))
sns.histplot(df["Life expectancy "], kde=True, bins=30)
plt.title("Distribution of Life Expectancy")
plt.xlabel("Life Expectancy (Years)")
plt.show()
```

Life Expectancy Over Years (Trend)

```python
avg_life_exp = df.groupby("Year")["Life expectancy "].mean().reset_index()

px.line(
    avg_life_exp,
    x="Year",
    y="Life expectancy ",
    title="Average Life Expectancy Over Years",
    template="plotly_dark"
)
```

Developed vs Developing Comparison

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x="GDP",
    y="Life expectancy ",
    hue="Status",
    data=df
)
plt.title("Life Expectancy vs GDP")
plt.show()
```

Life Expectancy vs GDP

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x="GDP",
    y="Life expectancy ",
    hue="Status",
    data=df
)
plt.title("Life Expectancy vs GDP")
plt.show()
```

Life Expectancy vs Schooling

```python
schooling_data = df.groupby("Schooling")["Life expectancy "].mean().reset_index()

px.line(
    schooling_data,
    x="Schooling",
    y="Life expectancy ",
    title="Life Expectancy vs Years of Schooling",
    template="plotly_dark"
)
```

Life Expectancy vs Alcohol Consumption

```python
px.scatter(
    df,
    x="Population",
    y="Life expectancy ",
    animation_frame="Year",
    hover_name="Country",
    title="Population vs Life Expectancy Over Time",
    template="plotly_dark"
)
```

Correlation Analysis

```python
numerical_cols = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(16,12))
sns.heatmap(numerical_cols.corr(), cmap="viridis", annot=False)
plt.title("Correlation Heatmap")
plt.show()
```

Outlier Detection (Boxplots)

```python
plt.figure(figsize=(18,14))
for i, col in enumerate(numerical_cols.columns):
    plt.subplot(5,5,i+1)
    sns.boxplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```
